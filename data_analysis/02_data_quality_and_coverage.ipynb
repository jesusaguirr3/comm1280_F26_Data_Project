## The warning sign

When we counted new construction permits by year, the numbers climbed very
fast in the late 2010s. That would mean Philadelphia went from almost no
building to a huge boom in three years. That seemed unlikely, so we tested it.

```python
import pandas as pd
import matplotlib.pyplot as plt

RAW = '../data/raw/'
NAVY = '#0B2545'
GOLD = '#C9A227'

by_year = pd.read_csv(RAW + 'all_permits_by_year_2016_2024.csv')
by_year
```

## The test

If building really grew, we should also see growth in the total number of
permits the city handed out. If only new construction grows while the total
stays flat, then nothing changed in the city and something changed in the
records.

```python
plt.figure(figsize=(10, 3.5))

plt.subplot(1, 2, 1)
plt.plot(by_year['year'], by_year['all_permits'], marker='o', color=NAVY)
plt.title('Every permit the city issued')
plt.ylim(0, 70000)
plt.grid(alpha=0.3)

plt.subplot(1, 2, 2)
plt.plot(by_year['year'], by_year['new_construction'], marker='o', color=GOLD)
plt.title('Permits labelled New Construction')
plt.ylim(0, 12000)
plt.grid(alpha=0.3)

plt.tight_layout()
plt.show()
```

The left chart is basically flat. The city issued somewhere between 44,000
and 64,000 permits every year, with a dip in 2020 for obvious reasons.

The right chart goes from 326 in 2016 to 10,281 in 2019. That is more than
thirty times as many, in a city whose total permit count did not move.

So the jump is a change in labelling, not a change in building.

## When did the labelling change?

Philadelphia switched to a new permit computer system. The old one wrote
categories in capitals, like `ALTERATION PERMIT`. The new one writes them
normally, like `Residential Building Permit`. If we count each style by month,
we can see exactly when the switch finished.

```python
systems = pd.read_csv(RAW + 'permit_record_system_by_month_2016_2024.csv')

# Just the months from 2019 on, so the chart is readable.
recent = systems[systems['month'] >= '2019-01']

plt.figure(figsize=(11, 3.5))
plt.bar(recent['month'], recent['old_system'], color=NAVY, label='Old system')
plt.bar(recent['month'], recent['new_system'], bottom=recent['old_system'],
        color=GOLD, label='New system')
plt.xticks(rotation=90)
plt.ylabel('Permits issued')
plt.legend()
plt.tight_layout()
plt.show()
```

```python
# Which months still contain old system records?
still_old = systems[systems['old_system'] > 0]

print('Last month with old system records:', still_old['month'].max())
```

## What we decided

The old system stops in **March 2020**. From April 2020 on, every permit comes
from the new system and is labelled the same way.

So our data starts in **April 2020**, and every number we report comes from
then onward.

### What this costs us

We wanted to compare against 2018 and 2019. We cannot. Those years are not
counted the same way as 2022 onward, and using them would have produced a big
fake result.

What we have left is April 2020 through December 2024. That is about twenty
months before the deadline and three full years after it, which is enough.
